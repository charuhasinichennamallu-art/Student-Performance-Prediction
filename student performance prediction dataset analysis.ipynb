```python
X_early = df.drop(columns=["G1", "G2", "G3"])
y = df["G3"]

print(X_early.shape)
print(y.shape)
```

```python
categorical_cols = X_early.select_dtypes(include="object").columns
numerical_cols = X_early.select_dtypes(exclude="object").columns

print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)
```

```python
categorical_cols = X_early.select_dtypes(include="str").columns
numerical_cols = X_early.select_dtypes(exclude="str").columns
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore")

encoded = encoder.fit_transform(X_early[categorical_cols])

print(encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_early,
    y,
    test_size=0.2,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", "passthrough", numerical_cols)
    ]
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("X_train_processed:", X_train_processed.shape)
print("X_test_processed:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train_processed, y_train)
```

```python
y_pred = model.predict(X_test_processed)

print(y_pred)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
from sklearn.metrics import mean_squared_error

rmse = mean_squared_error(y_test, y_pred) ** 0.5

print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R²:", r2)
```

```python
comparison = pd.DataFrame({
    "Actual G3": y_test.values,
    "Predicted G3": y_pred
})

print(comparison.head(10))
```

```python
X_progress = df.drop(columns=["G3"])
y = df["G3"]

print(X_progress.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_progress,
    y,
    test_size=0.2,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
categorical_cols_progress = X_progress.select_dtypes(include="str").columns
numerical_cols_progress = X_progress.select_dtypes(exclude="str").columns

print("Categorical:", len(categorical_cols_progress))
print("Numerical:", len(numerical_cols_progress))
```

```python
categorical_cols_progress = X_progress.select_dtypes(include="str").columns
numerical_cols_progress = X_progress.select_dtypes(exclude="str").columns

preprocessor_progress = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols_progress),
        ("num", "passthrough", numerical_cols_progress)
    ]
)

X_train_progress, X_test_progress, y_train_progress, y_test_progress = train_test_split(
    X_progress,
    y,
    test_size=0.2,
    random_state=42
)

X_train_progress_processed = preprocessor_progress.fit_transform(X_train_progress)
X_test_progress_processed = preprocessor_progress.transform(X_test_progress)

print("X_train_progress_processed:", X_train_progress_processed.shape)
print("X_test_progress_processed:", X_test_progress_processed.shape)
```

```python
from sklearn.linear_model import LinearRegression

model_progress = LinearRegression()

model_progress.fit(
    X_train_progress_processed,
    y_train_progress
)
```

```python
y_pred_progress = model_progress.predict(X_test_progress_processed)

print(y_pred_progress)
```

```python
from sklearn.metrics import mean_absolute_error

mae_progress = mean_absolute_error(
    y_test_progress,
    y_pred_progress
)

print("Model 2 MAE:", mae_progress)
```

```python
from sklearn.metrics import mean_squared_error

rmse_progress = mean_squared_error(
    y_test_progress,
    y_pred_progress
) ** 0.5

print("Model 2 RMSE:", rmse_progress)
```

```python
from sklearn.metrics import r2_score

r2_progress = r2_score(
    y_test_progress,
    y_pred_progress
)

print("Model 2 R²:", r2_progress)
```

```python
#random forest regressor

from sklearn.ensemble import RandomForestRegressor
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
rf_model.fit(
    X_train_progress_processed,
    y_train_progress
)
```

```python
y_pred_rf = rf_model.predict(X_test_progress_processed)

print(y_pred_rf)
```

```python
from sklearn.metrics import mean_absolute_error

mae_rf = mean_absolute_error(
    y_test_progress,
    y_pred_rf
)

print("Random Forest MAE:", mae_rf)
```

```python
from sklearn.metrics import mean_squared_error

rmse_rf = mean_squared_error(
    y_test_progress,
    y_pred_rf
) ** 0.5

print("Random Forest RMSE:", rmse_rf)
```

```python
from sklearn.metrics import r2_score

r2_rf = r2_score(
    y_test_progress,
    y_pred_rf
)

print("Random Forest R²:", r2_rf)
```
